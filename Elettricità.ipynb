** Access to Electricity**

Il dataset Access to Electricity fornisce dati sulla quota di popolazione con accesso all’energia elettrica a livello globale, nazionale e regionale. Le informazioni sono generalmente espresse come percentuale sul totale della popolazione, con una copertura storica pluriennale che consente l’analisi dei trend nel tempo.

**Contenuto del dataset**

Il dataset include tipicamente:

Anno di riferimento

Paese o area geografica

Percentuale della popolazione con accesso all’elettricità

(Opzionale) Dati distinti per aree urbane e rurali

**Finalità e utilizzo**

Il dataset è progettato per supportare:

analisi sull’evoluzione dell’accesso ai servizi energetici;

valutazioni del divario tra Paesi sviluppati e Paesi in via di sviluppo;

analisi territoriali delle differenze urbano–rurali;

ricerche in ambito di sviluppo economico, infrastrutture e politiche energetiche.

1 nuova colonna creata usando apply e lambda (Questa riga crea una nuova colonna chiamata Rounded_2016_Access e fa:

“Per ogni valore della colonna 2016, se è un numero vero, arrotondalo;
Se invece è vuoto (NaN), lascialo così com'è.”)

```python
df['2016 [YR2016]'] = pd.to_numeric(df['2016 [YR2016]'], errors='coerce')

df['Rounded_2016_Access'] = df['2016 [YR2016]'].apply(lambda x: round(x) if pd.notna(x) else x)
df.head()
```

5 grafici comprensivi di titolo, labels e interpretazione (commento per ciascuno) di cui
almeno 1 boxplot, 1 istogramma e 1 scatterplot usando sia Matplotlib che Seaborn.

Commento: Questo grafico mostra l'andamento della percentuale di accesso
all'elettricità in Australia dal 2012 al 2016. Si può osservare
un trend
costante al 100%, indicando una copertura completa in questo periodo

```python
year_columns = [col for col in df.columns if 'YR' in col]
for col in year_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df_melted = df.melt(id_vars=['Country Name', 'Continent'],
                    value_vars=year_columns,
                    var_name='Year_Raw',
                    value_name='Access_to_Electricity')

df_melted['Year'] = df_melted['Year_Raw'].str.extract(r'(\d{4})').astype(int)

australia_data = df_melted[df_melted["Country Name"] == "Australia"].sort_values(by='Year')

plt.figure(figsize=(10,6))
plt.plot(australia_data["Year"], australia_data["Access_to_Electricity"], marker='o')
plt.title("Accesso all'Elettricità in Australia nel Tempo")
plt.xlabel("Anno")
plt.ylabel("Percentuale di Accesso")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Commento: Questo istogramma mostra la distribuzione della percentuale di accesso
all'elettricità tra i vari paesi nel 2016. Si possono osservare raggruppamenti
di paesi con livelli di accesso simili, evidenziando la variabilità globale.

```python
data_2016_access = df_melted[df_melted['Year'] == 2016]['Access_to_Electricity'].dropna()

plt.figure(figsize=(10,6))
plt.hist(data_2016_access, bins=20, edgecolor='black')
plt.title("Distribuzione dell'Accesso all'Elettricità nel 2016 (tutti i paesi)")
plt.xlabel("Percentuale di Accesso")
plt.ylabel("Numero di Paesi")
plt.grid(True)
plt.tight_layout()
plt.show()
```

Commento: Questo scatter plot visualizza la percentuale di accesso all'elettricità
per ciascun paese nel 2016. Permette di identificare rapidamente paesi con
livelli di accesso molto alti o molto bassi e la variabilità tra le nazioni.

```python
plt.figure(figsize=(15, 10))
plt.title("Accesso all'Elettricità nel 2016 per Paese")

data_for_scatterplot_2016 = df_melted[df_melted['Year'] == 2016].dropna(subset=['Access_to_Electricity'])
sns.scatterplot(data=data_for_scatterplot_2016, x='Country Name', y='Access_to_Electricity')

plt.ylabel("Percentuale di Accesso")
plt.xticks(rotation=90, fontsize=8)
plt.grid(True)
plt.tight_layout()
plt.show()

# Commento: Questo scatter plot visualizza la percentuale di accesso all'elettricità
# per ciascun paese nel 2016. Permette di identificare rapidamente paesi con
# livelli di accesso molto alti o molto bassi e la variabilità tra le nazioni.
```

-Commento: Questo scatter plot confronta la percentuale di accesso all'elettricità
-per i paesi in Europa e Asia nel 2016. Ogni punto rappresenta un paese,
-ed è colorato in base al suo continente. Questo grafico permette di visualizzare:
-la variabilità all'interno di ciascun continente e tra i due.

Filtra il DataFrame fuso per l'anno 2016 e solo per Europa e Asia.


```python
data_europe_asia_2016 = df_melted[
    (df_melted['Year'] == 2016) &
    (df_melted['Continent'].isin(['Europe', 'Asia']))
].dropna(subset=['Access_to_Electricity'])

plt.figure(figsize=(18, 8)) # Adjust figure size for better readability
plt.title("Accesso all'Elettricità nel 2016: Confronto tra Europa e Asia")

sns.scatterplot(data=data_europe_asia_2016,
                x='Country Name',
                y='Access_to_Electricity',
                hue='Continent',
                s=100) # s for marker size

plt.xlabel("Paese")
plt.ylabel("Percentuale di Accesso")
plt.xticks(rotation=90, fontsize=8)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.hist(data_2016_access, bins=20, edgecolor='black')
plt.title("Distribuzione dell'Accesso all'Elettricità nel 2016")
plt.xlabel("Percentuale di Accesso")
plt.ylabel("Frequenza")
plt.grid(True)
plt.tight_layout()
plt.show()

# Commento: Questo istogramma mostra la distribuzione della percentuale di accesso
# all'elettricità tra i vari paesi nel 2016, usando i dati preparati in precedenza.
# Si possono osservare raggruppamenti di paesi con livelli di accesso simili.
```

Commento: Questo grafico a barre mostra la media della percentuale di accesso
all'elettricità per ciascun paese nel periodo 2012-2016. I paesi sono ordinati
per facilitare il confronto dei livelli medi di accesso.

```python
mean_country = df_melted.groupby("Continent")["Access_to_Electricity"].mean().sort_values()

plt.figure(figsize=(15,8))
plt.bar(mean_country.index, mean_country.values)
plt.title("Media Accesso all'Elettricità (2012-2016) per Paese")
plt.xlabel("Paese")
plt.ylabel("Media Percentuale di Accesso")
plt.xticks(rotation=90, fontsize=8)
plt.tight_layout()
plt.show()
```

Andamento globale accesso all'elettricità, nel tempo, dal 2012 al 2016

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Year', y='Access_to_Electricity', data=df_melted)
plt.title("Distribuzione dell'Accesso all'Elettricità nel Mondo per Anno")
plt.xlabel("Anno")
plt.ylabel("Percentuale di popolazione con accesso")
plt.grid(True)
plt.show()
```
